```python
import pandas as pd

# Load the data - change the filename if yours is different!
df = pd.read_csv('student_mental_health_burnout_1M.csv')

# Let's see the columns and the first few rows
print(df.columns)
df.head()
```

```python
# 1. Check the "Health" of the data
print("--- Missing Values ---")
print(df.isnull().sum())

print("\n--- Summary Statistics ---")
display(df.describe())

# 2. Visualize the Target Variable (Burnout)
import seaborn as sns
import matplotlib.pyplot as plt

# Replace 'Burnout_Rate' with the actual column name from your CSV
# Common names in this dataset: 'Burnout Rate', 'Burnout Score'
target_col = 'burnout_score' 

plt.figure(figsize=(10, 6))
sns.histplot(df[target_col], kde=True, color='salmon')
plt.title('Distribution of Student Burnout')
plt.show()
```

```python
# Check data types
print(df.dtypes)

# If 'academic_year' or 'exam_pressure' are text, we can convert them:
# df['exam_pressure'] = df['exam_pressure'].map({'Low': 0, 'Medium': 1, 'High': 2})
```

```python
# Select only numeric columns for correlation
numeric_df = df.select_dtypes(include=['float64', 'int64'])

plt.figure(figsize=(12, 8))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('What correlates with Burnout?')
plt.show()
```
